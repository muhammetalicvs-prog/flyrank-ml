## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/muhammetalicvs-prog/flyrank-ml/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print("Total rows:", len(df))
print("Total columns:", df.shape[1])
print("Unique clients:", df["client_id"].nunique())

declining_pages = (df["trend_direction"] == "down").sum()
declining_rate = (df["trend_direction"] == "down").mean() * 100

stale_visible_pages = (
    (df["days_since_last_update"] >= 180) &
    (df["impressions_90d"] >= 500)
).sum()

print("Declining pages:", int(declining_pages))
print("Declining rate:", round(declining_rate, 2), "%")
print("Stale but visible pages:", int(stale_visible_pages))
```

The starter dataset contains many pseudonymized content pages and several useful performance signals. The numbers above show that there are declining pages and stale pages that still receive search visibility.

This makes the lane worth exploring because the useful task is not only to predict a label. The real task is to rank pages so that a reviewer can decide where to spend limited time first.
